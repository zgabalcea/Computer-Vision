```python

#*Check if the actual rectangle, intersects more than 30% with another one
def rectangle_intersects_polygons_or_is_near(rectangle, polygons):
    
    for p in polygons:
        if rectangle.intersects(p):
            # print("intersection of polygons distance : ", rectangle.distance(p))
            return True
        
    for p in polygons:
        if rectangle.distance(p) <= 20:
            polygons.append(rectangle)
            return True
        
    return False
```
